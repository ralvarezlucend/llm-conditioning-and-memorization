```python
import pandas as pd

batch_size = 64
repeated_pool_ratio = 0.1
repeated_batch_ratios = [0.1, 0.2, 0.4]
dataset_sizes = [100, 500, 1000, 1500, 2000]

experiments = []
for dataset_size in dataset_sizes:
    for repeated_batch_ratio in repeated_batch_ratios:
        n_repeated_pool = int(dataset_size * repeated_pool_ratio)
        n_repeated_batch = int(batch_size * repeated_batch_ratio)

        if n_repeated_batch < n_repeated_pool:
            experiments.append(
                (dataset_size, n_repeated_pool, n_repeated_batch, repeated_batch_ratio)
            )

df = pd.DataFrame(
    experiments,
    columns=[
        "dataset_size",
        "n_repeated_pool",
        "n_repeated_batch",
        "repeated_batch_ratio",
    ],
)
df
```
