```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

cmap = plt.get_cmap("coolwarm")
norm = mcolors.Normalize(vmin=150, vmax=500)

fig, ax = plt.subplots(figsize=(8, 0.5))

cb = plt.colorbar(
    plt.cm.ScalarMappable(norm=norm, cmap=cmap),
    cax=ax,
    orientation="horizontal",
)

# Tick size
cb.ax.tick_params(labelsize=16)

# Title at the top
fig.suptitle("Information Density (in bits)", fontsize=20, y=1.65)

# plt.tight_layout()
plt.savefig('plots/information-density-cbar.svg', format="svg", bbox_inches="tight")
```
